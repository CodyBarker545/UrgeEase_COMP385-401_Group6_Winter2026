```python
import kagglehub
import shutil
from pathlib import Path

BASE_DIR = Path.cwd()
DATA_DIR = BASE_DIR / "data"
DATA_DIR.mkdir(parents=True, exist_ok=True)

datasets = {
    "social_media_addiction_relationships.csv": "adilshamim8/social-media-addiction-vs-relationships",
    "social_media_users.csv": "souvikahmed071/social-media-and-mental-health"
    
}

for filename, dataset_id in datasets.items():

    print(f"\nDownloading {dataset_id}...")

    download_path = kagglehub.dataset_download(dataset_id)

    download_path = Path(download_path)

    # find the CSV file inside the downloaded dataset
    csv_file = list(download_path.rglob("*.csv"))[0]

    target_file = DATA_DIR / filename

    shutil.copy(csv_file, target_file)

    print(f"Saved as: {target_file}")

print("\nAll datasets downloaded successfully.")
```

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("data")

for file in DATA_DIR.glob("*.csv"):
    print("Dataset:",file.name)
    print(f"Processing {file}...")
    
    df = pd.read_csv(file)
    
    print("Shape:", df.shape)
    print("Columns:")
    
    for col in df.columns:
        print(" -", col)
```
